## Bagian 1: Penerapan Computational Thinking (Kasus Kebun Ibu Mira)

### Studi Kasus:
Ibu Mira adalah petani di musim kering yang membutuhkan pasokan air minimal **100 mm per minggu** untuk kebunnya. Kita akan membantu Ibu Mira menghitung total curah hujan selama 7 hari, mengecek apakah target terpenuhi, dan menghitung berapa kekurangannya jika air tidak cukup.

###1.1 Decomposition (Dekomposisi)
Memecah masalah besar menjadi bagian-bagian kecil yang lebih mudah dikelola.

Untuk menyelesaikan masalah Ibu Mira, kita memecahnya menjadi:
* Input Data: Mencatat curah hujan harian selama 7 hari.
* Kalkulasi: Menghitung total (sum) dari data curah hujan tersebut.
* Evaluasi: Membandingkan total dengan ambang batas (threshold) 100 mm.
* Output: Menghitung selisih kekurangan jika target tidak tercapai.

###1.2 Pattern Recognition (Pengenalan Pola)
Mencari kesamaan atau tren dalam data. Dalam kasus ini, kita mengenali pola berikut:
* Setiap hari memiliki struktur data yang sama (angka dalam satuan mm).
* Proses penjumlahan dilakukan berulang untuk setiap elemen data (iterasi).
* Logika pengecekan selalu sama: Jika Total < Target, maka Kurang.

###1.3 Abstraction (Abstraksi)
Fokus pada informasi penting dan mengabaikan detail yang tidak relevan. Kita mengabaikan detail seperti:
* Jenis tanaman Ibu Mira.
* Waktu spesifik hujan (pagi atau sore).
* Alat ukur yang digunakan.

**Variabel Penting yang Kita Simpan:**
* `curah_hujan_harian` (List angka)
* `target_minimal` (Konstanta: 100)
* `total_hujan` (Hasil jumlah)

###1.4 Algorithm (Algoritma)
Menyusun langkah-langkah sistematis untuk menyelesaikan masalah.
* Mulai.
* Siapkan Wadah: Buat tempat penyimpanan untuk `total_hujan` dan beri nilai awal 0.
* Input Data: Masukkan 7 angka curah hujan harian ke dalam daftar `curah_hujan_harian`.
* Hitung Akumulasi: Tambahkan setiap angka dari daftar ke dalam `total_hujan`.
* Cek Target: Bandingkan hasil `total_hujan` dengan angka 100 (`target_minimal`).
* Percabangan Logika:
  * JIKA `total_hujan` mencapai 100 atau lebih:
    * Tampilkan pesan: "Pasokan air cukup".
  * JIKA TIDAK (kurang dari 100):
    * Hitung Kekurangan = `target_minimal - total_hujan`.
    * Tampilkan pesan: "Pasokan air kurang" dan informasikan nilai Kekurangan-nya.
* Selesai.

#### **Good variable**
Nama yang baik mendeskripsikan apa yang disimpan oleh variabel tersebut, bukan sekadar tipe datanya.

Nama yang buruk membuat kode lebih sulit untuk diperiksa (debugging) dan dipelihara.

* Misterius: Hindari nama seperti `jml` (`jumlah`), `psn` (`pesan`), atau `nm_pengguna` (`nama_pengguna`).

* Ambigu: Hindari penggunaan huruf kecil `l`, huruf besar `O`, atau huruf besar `I` sebagai nama satu karakter karena mereka sangat mirip dengan angka '1' dan '0'.

* Nama yang Terlalu Umum: Nama seperti data, nilai, temp, atau sesuatu tidak memberikan konteks sama sekali.

* Nama yang Redundan: Hindari memasukkan tipe data ke dalam nama, seperti `nama_string` atau `usia_int`.

### 3.2 Dictionary (dict)

`Dictionary` merupakan struktur data seperti `list`. Dalam `list`, indeks merupakan bilangan bulat, namun di dalam `dictionary` indeks bisa (hampir) tipe data apa saja.

`Dictionary` juga bisa dianggap sebagai pemetaan antara sebuah indeks yang disebut **keys** dan sebuah **value**. Setiap key **memetakan** ke suatu value. Hubungan key dan value ini disebut **key-value pair**.

![image.png](https://www.learnbyexample.org/wp-content/uploads/python/Dictionary-Key-Value-Pairs-Illustration.png)


[sumber](https://www.learnbyexample.org/python-dictionary/)

Ada berbagai macam method pada `dictionary` yang bisa digunakan. Pada pertemuan ini hanya akan dikenalkan pada method yang sering digunakan saja.

Contoh:


```
dict_1 = {'Nama': 'Farhan', 'Usia':24}
dict_2 = {'Nama': 'Tania', 'Usia':19}
```

|Fungsi|Method|Penggunaan|Output|
|-|-|-|-|
|mengetahui panjang dictionary|`len()`|`len(dict_1)`|`2`
|mendapatkan element dari key|`get(NAMA_KEY)`|`dict_1.get('Nama')`|`Farhan`|
|mendapatkan semua key |`keys()`|`dict_1.key()`|`['Nama', 'Usia']`|
|mendapatkan semua value|`values()`|`dict_1.values()`|`['Farhan', 19']`
|mengeluarkan atau menghapus key dan value |`pop()`| `dict_1.pop('Usia')`|`{'Nama':'Farhan'}`|
|menambah atau menggabungkan data|`update()`|`dict_1.update(dict_2)`|`{'Nama': 'Farhan', 'Usia':24, 'Nama': 'Tania', 'Usia':19}`
|membuat duplikasi|`copy()`|`dict_1.copy()`|duplikasi dari dict_1
|menghapus semua key dan value|`clear()`|`dict_1.clear`|`{}`|


````python
nilai = {
    "nama": "Laila",
    "umur": 28,
    "aktif": True
}

nilai["nama"]   # "Laila"
nilai["umur"]   # 28
print(nilai["nama"])
````

### 3.4 Set

**Set** adalah sebuah struktur data yang memungkinkan kita untuk menyimpan sebuah data dengan tidak berurutan **(unordered)**. Setiap element di dalam set adalah **unique** (tidak ada duplikat).

**Set** ini bisa berisi tipe data seperti `string`, `integer`, `float`, bahkan `tuple`. Namun di dalam set tidak diperbolehkan untuk menyimpan tipe data seperti `list`, `set`, atau `dictionary` sebagai sebuah elements

Sebuah `set` dalam python di representasikan dengan **tanda kurung kurawal** `{}` dan dipisahkan oleh **tanda koma** `,`

````python
# coba jalankan code di bawah
set_angka = {1, 2, 3, 3, 1, 4}
set_buah = {'Semangka', 'Alpukat', 'Apel', 'Semangka', 'Apel'}

print(set_angka)
print(set_buah)
````

Contoh:


```
set_1 = {3, 4, 5, 6}
set_2 = {1, 2, 3, 6}
```

|Fungsi|Method|Penggunaan|Output|
|-|-|-|-|
|mengetahui panjang set|`len()`|`len(set_1)`|`4`|
|melihat apakah item ada di dalam set atau tidak|`in`| `3 in set_1`|`True`|
|menambah element|`add(value)`|`set_1.add(7)`|`{3, 4, 5, 6, 7}`|
|menambah set, list atau tuple ke dalam set|`update(value)`|`set_1.update(set_2)`|`{1, 2, 3, 4, 5, 6}`

Struktur perulangan `while` adalah sebagai berikut

```python
while kondisi:
    Statements

    Statements
```
- keyword `while` menandakan awal dari `while` loop.
- `kondisi` adalah ekspresi yang dievaluasi setiap kali sebelum perulangan dijalankan selama bernilai `True`, perulangan akan terus berjalan.

````python
# Versi FOR
for numb in range(1, 6): #range(start, stop)
    print(numb)

"""
range goes from start up to but not including stop.
This is called exclusive end.
"""

# Versi WHILE
angka = 1
while angka <= 5:
    print(angka)
    angka += 1  # ← ini yang bikin kondisi akhirnya False
````

````python
buah = ["apel", "mangga", "jeruk"]

# Versi FOR
for item in buah:
    print(item)

# Versi WHILE
i = 0
while i < len(buah):
    print(buah[i])
    i += 1

"""
💡 Perhatikan — versi while butuh variabel i yang dikelola manual.
Versi for jauh lebih bersih untuk kasus seperti ini.
"""
````

- Tahu berapa kali mau ulang?  →  pakai FOR
- Tidak tahu, tergantung kondisi?  →  pakai WHILE

````python
# Break -> Hentikan loop paksa
# Berhenti ketika ketemu angka 3
print ("contoh break")
for i in range(1, 6):
    if i == 3:
        break
    print(i)

# Continue -> Lewati iterasi ini, lanjut ke berikutnya
# Lewati angka 3, lanjutkan sisanya
print ("contoh continue")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

# Pass -> Tidak melakukan apa-apa
# pass dipakai sebagai placeholder
print ("contoh pass")
for i in range(1, 6):
    if i == 3:
        pass       # ← tidak ada efeknya, kode lanjut normal
    print(i)
````

#### Latihan For dan While - Break, Continue dan Pass

````python
# 1. Perulangan 'for' (Meng-iterasi riwayat curah hujan mingguan)
riwayat_hujan = [12, 18, 5, 22, 9, 14, 30]

for index, curah_hujan in ...(riwayat_hujan):
    if curah_hujan < 0: # CONTINUE: data tidak valid
        print(f"Hari ke-{index+1}: data tidak valid, dilewati.")
        ...

    elif curah_hujan == 0: # PASS
        ...  # nanti bisa diisi: kirim notifikasi "hari tanpa hujan"

    elif curah_hujan > 25: # BREAK: curah hujan ekstrem
        print(f"Hari ke-{index+1}: {curah_hujan} mm -> EKSTREM! Analisis dihentikan.")
        ...

    elif curah_hujan >= 15: # Tinggi
        print(f"Hari ke-{index+1}: {curah_hujan} mm -> Status Hujan: Tinggi")

    else: # Rendah (< 15)
        print(f"Hari ke-{index+1}: {curah_hujan} mm -> Status Hujan: Rendah")
print("-" * 40)

# 2. Perulangan 'while' (Simulasi pengosongan tangki air dari 50 liter ke 0 liter)
tangki_air = 50
while ... > 0:

    if tangki_air < 0: # BREAK: kebocoran
        print("Kebocoran terdeteksi! Pengairan darurat dihentikan.")
        break

    ... tangki_air == 20: # CONTINUE: cadangan
        print(f"Volume tangki air saat ini: {tangki_air} liter. Cadangan! Pengairan dilewati.")
        tangki_air -= 10
        continue

    ...: # Normal: alirkan air
        print(f"Volume tangki air saat ini: {tangki_air} liter. Mengalirkan air ke kebun...")
        tangki_air -= 10

print("Tangki air kosong! Pengairan selesai.")
````

Dalam sesi ini, kita akan membahas apa itu error dan bagaimana cara menghadapinya. Error adalah sesuatu yang sangat lumrah dijumpai dalam proses pembuatan program. Bahkan, justru dari error itulah pemahaman kita terhadap kode yang ditulis semakin berkembang.
Saat menghadapi error, ada beberapa langkah yang bisa dicoba:

* Jika masalah belum teratasi, ulangi proses pencarian dengan kata kunci yang berbeda

Adapun beberapa jenis error yang paling sering ditemui antara lain <b> Syntax Error, Indentation Error, dan Name Error</b>.

#### Syntax Error

* Syntax Error terjadi ketika kode yang ditulis tidak sesuai dengan aturan penulisan yang berlaku di Python.

#### Try & Except

Setelah memahami berbagai jenis error, sekarang kita akan belajar cara menanganinya secara lebih elegan menggunakan `try` dan `except`. Blok `try-except` memungkinkan program tetap berjalan meskipun terjadi error alih-alih langsung berhenti, program akan menjalankan instruksi alternatif yang sudah kita siapkan di dalam blok `except`.

```python
try:
    # kode yang mungkin menyebabkan error
except:
    # kode yang dijalankan jika terjadi error
```

````python
# Menangani input yang tidak valid
try:
    nilai = int(input("Masukkan nilai: "))
    print(f"Nilai kamu adalah {nilai}")
except:
    print("Input tidak valid! Harap masukkan angka.")

"""
Jika user memasukkan huruf, misal "abc":
Input tidak valid! Harap masukkan angka.
Tanpa try-except, program akan langsung crash dengan ValueError.
"""
````

Jenis error dapat di explore di https://docs.python.org/3/tutorial/errors.html

````python
# Menangkap jenis error secara spesifik
try:
    angka = int(input("Masukkan angka: "))
    hasil = 100 / angka
    print(f"Hasilnya: {hasil}")
except ValueError:
    print("Error: Input harus berupa angka!")
except ZeroDivisionError:
    print("Error: Tidak bisa membagi dengan angka 0!")
````
